```python
import pandas as pd

df = pd.read_csv("../data/raw/test_log.csv", header=None, names=["timestamp_us", "value"])
deltas = df["timestamp_us"].diff()

print(deltas.describe())
print("Effective rate (Hz):", 1e6 / deltas.mean())
```

```python
import pandas as pd

df = pd.read_csv("../data/raw/test_log.csv")
deltas = df["timestamp_us"].diff()

print(deltas.describe())
print("Effective rate (Hz):", 1e6 / deltas.mean())
```

```python
import numpy as np
import matplotlib.pyplot as plt

values = np.loadtxt("../data/raw/isometric_hold_01.csv", skiprows=1)
t = np.arange(len(values)) / 1000   # seconds; valid because the rate is a verified 1000 Hz

plt.figure(figsize=(12, 4))
plt.plot(t, values, linewidth=0.5)
plt.xlabel("Time (s)")
plt.ylabel("ADC value")
plt.title("Raw sEMG - isometric hold")
plt.show()

print(len(values), "samples,", t[-1], "s")
```
